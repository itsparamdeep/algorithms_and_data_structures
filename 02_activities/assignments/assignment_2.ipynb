## Problem Two: Path to Leaves
**Description**  
Given the `root` of a binary tree, return all root to leaf paths in any order.

**Examples**

* Example 1:

    ![](./images/q1_ex1.png)

    Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

    Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

* Example 2:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 7, 8]`

    Output: [[10, 7], [10, 9, 8]]

**Problem 2 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# A binary tree is a tree data structure in which each node has at most two children. These children are typically distinguished as the left child and the 
# right child. Each node in a binary tree contains a value, and it may have references to its left and right children. The left child is a reference to another 
# node that is considered the left descendant of the current node, while the right child is a reference to another node that is considered the right descendant of
# the current node. Binary trees are commonly used in computer science for various applications, including searching, sorting, and representing hierarchical data
# structures.

class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right


from typing import List

def bt_path(root: "TreeNode") -> List[List[int]]:
    result = []
    if root is None:
        return result

    def dfs(node, path):
        if node is None:
            return
        path.append(node.val)
        if node.left is None and node.right is None:
            result.append(list(path))
        else:
            dfs(node.left, path)
            dfs(node.right, path)
        path.pop()

    dfs(root, [])
    return result
```


4. Copy the solution your partner wrote. 


```python
# Your answer here
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    result = []
    zeros = 0

    for num in nums:
        if num == 0:
            zeros += 1
        else:
            result.append(num)

    result.extend([0] * zeros)
    return result
```
